```python
##Code here

# Checks every cell on the DataFrame and creates a new table of True and False values showing where data is missing.
missing_values = df.isnull()

# As shown on the table, there are no missing values as it returns all cells are false
display(missing_values)
```

```python
# Another way to check for missing values, this filters your DataFrame to keep and display only the rows that have at least one missing value.
rows_with_nulls = df[missing_values.any(axis=1)]

display(rows_with_nulls)
```

```python
# Summary statistics of numeric columns (age, salary, work_experience)
numeric_summary = df[['age', 'salary', 'work_experience']].describe().round(2)

display(numeric_summary)
# count: the number of non-missing(non-null)values
# mean: the average value
# std: standard deviation-how spread out the numbers are around the mean
# min: the lowest value in the column
# 25%: the 1st quartile- 25% of values fall below this number
# 50%: the median(middle value)- half the data is above, half is below
# 75%: the 3rd quartile- 75% of values fall below this number
# max: the highest value in the column
```

Task 3: Data Transformation
Objective: Perform basic data transformations such as creating new columns and renaming existing ones.

Instructions:

Create a new column called years_to_retirement by subtracting the age from 65.

Rename the salary column to annual_salary.

Sort the DataFrame based on work_experience in descending order.

```python
##Code here

# Adding a new column called years_to_retirement that calculates how many years each person has left until they reach the retirement age of 65
df['years_to_retirement'] = 65 - df['age']

display(df)
```

```python
# Renaming the salary column to annual_salary
df = df.rename(columns={'salary': 'annual_salary'})

display(df)
```

```python
# Sorting the DataFrame based on work_experience in descending order
df_sorted = df.sort_values(by='work_experience', ascending=False)

display(df_sorted)
```

Task 4: Handling Missing Data
Objective: Handle missing data (if any).

Instructions:

Introduce some missing values in the annual_salary column.

Handle the missing data by either filling it with the mean salary or dropping the rows with missing values.

```python
##Code here

# How to introduce some missing values in the annual_salary column.

# Numpy provides np.nan, which is the tool used to create missing values in Pandas.
import numpy as np

# Set rows 1 and 4 of 'annual_salary' to missing (Nan)
df.loc[[1, 4], 'annual_salary'] = np.nan

display(df)
```

```python
# Calculate average (mean) salary of the existing data (pandas automatically ignores Null/NaNs when calculating mean)
mean_salary = df['annual_salary'].mean()

# Fill Null/NaNs with the mean (average value)
df['annual_salary'] = df['annual_salary'].fillna(mean_salary)

display(df)
```

```python
# Dropping the rows with missing data

# First, let's create missing values based on a different category, specifically for employees under 25 years old.
df.loc[df['age'] < 25, 'annual_salary'] = np.nan

# Under 25 of age are employee_id=1 John (HR) and employee_id=2 Jack (Engineering)
display(df)
```

```python
# Drop only the rows where 'annual_salary' is Null/NaN
df = df.dropna(subset=['annual_salary'])

# John and Jack's row was dropped
display(df)
```

Task 5: Data Filtering and Selection
Objective: Select specific data based on conditions.

Instructions:

Filter the data to show only employees who are older than 30.

Select only the columns name, age, and annual_salary.

```python
##  Code here

# This filters the DataFrame to keep only emplyees who are age over 30 and we ascend them into order.
over_30 = df[df['age'] > 30].sort_values(by='age', ascending=True)

display(over_30)
```

```python
# Select only columns name, age, and annual_salary.
df_selected = df[['name', 'age', 'annual_salary']]

display(df_selected)
```

Task 6: Data Aggregation
Objective: Perform group-based aggregation.

Instructions:

Group the data by department and calculate the average salary for each department.

```python
## Code here

# Group all employees by their department, calculate the average salary for each group in descending order.
avg_salary_df = df.groupby('department')['annual_salary'].mean().round(2).reset_index().sort_values(by='annual_salary', ascending=False)

display(avg_salary_df)
```
